```python
import numpy as np
import matplotlib.pyplot as plt

Nc = 2.86e19
k = 8.62e-5
T = 300

ec_ef = np.linspace(0.01, 0.5, 1000)

n_total = Nc * np.exp(-ec_ef / (k * T))

plt.plot(ec_ef, n_total, linewidth=2)

plt.title(fr"Concentração de portadores $cm^{-3}$ em função de $E_c - E_f$ (eV)")
plt.xlabel(r"$E_c - E_f$ (eV)")
plt.ylabel(r"Concentração de portadores $cm^{-3}$")
plt.yscale('log')
plt.grid()
```

```python
Nd = 1e16
T = np.linspace(20, 700)
ec_ef = - k * T * np.log(Nd/Nc)

plt.plot(T, ec_ef, linewidth=2)

plt.title(fr"$E_c - E_f$ (eV) em função de T para silicio dopado com $N_D = 10^{16} cm^{-3} $")
plt.ylabel(r"$E_c - E_f$ (eV)")
plt.xlabel(r"T (K)")
plt.grid()
```
